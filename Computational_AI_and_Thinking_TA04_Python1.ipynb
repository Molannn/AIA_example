> **A variable is a name used to represent a value.**

```python

```

> **白話說明：** 變數就像**貼在東西上的名牌**。`a = 1` 的意思是「把 a 這張名牌貼到 1 上面」。
> 注意：這裡的 `=` **不是數學的等於**，而是「把右邊的值交給左邊的名字」。

We can use this diagram to understand the relationship between variables and values:

# if, else and elif

## Indentation 縮排
Python uses indentation (spaces at the start of a line) to group code into blocks.
```python
if something:
    print('123123')
    print('testing')
else:
    print('else something')
    print('same block in testing')
```
In the example above, if the condition is True, Python runs the two `print` lines under `if`, because they share the same indentation.
Otherwise, it runs the two `print` lines under `else`.

> **白話說明：**
> - 縮排就像**文章的段落**：往內縮的那幾行，都屬於上面那句 `if` 或 `else`。
> - 縮排通常是 **4 個空格**（Colab 按 `Tab` 也會自動變成空格）。
> - `if`、`elif`、`else` 那一行的最後**一定要加冒號 `:`**，忘記加是最常見的錯誤。

## Using if statements

An `if` statement lets the program make decisions: **if** something is true, **then** do something.

```python
>>> its_raining = True
>>> if its_raining:
...     print("It's raining!")
...
It's raining!
>>> its_raining = False
>>> if its_raining:
...     print("It's raining!")        # nothing happens
...
>>>
```

## Using else

When the `if` condition is not met, the program runs the code under `else`.

```python
its_raining = False
if its_raining:
    print("It's raining!")
else:
    print("It's not raining.")
```
```
It's not raining.
```

> **白話說明：** `if ... else ...` 就是「如果……就……，不然就……」。兩條路只會走其中一條。

````python
its_raining = True
if its_raining:
    print("It's raining!")
else:
    print("It's not raining.")
# 試試看把第一行改成 False 再執行一次
````

## elif

If we have multiple conditions to check, we might write code like this:

```python
print("Hello!")
word = input("Enter something: ")

if word == "hi":
    print("Hi to you too!")
else:
    if word == "hello":
        print("Hello hello!")
    else:
        if word == "howdy":
            print("Howdyyyy!")
        else:
            if word == "hey":
                print("Hey hey hey!")
            else:
                print("I don't know what", word, "means.")
```

It looks messy. `elif` (short for "else if") solves this neatly:

```python
print("Hello!")
word = input("Enter something: ")

if word == "hi":
    print("Hi to you too!")
elif word == "hello":
    print("Hello hello!")
elif word == "howdy":
    print("Howdyyyy!")
elif word == "hey":
    print("Hey hey hey!")
else:
    print("I don't know what", word, "means.")
```

> **白話說明：** `elif` 就像**選擇題**：Python 由上往下一個一個檢查，**一碰到成立的就執行那一段，然後整組結束**，後面的不會再看。全部都不成立，才會跑 `else`。
>
> 另外，`print()` 裡用逗號 `,` 隔開好幾個東西，印出來時會自動在中間加空格。

````python
print("Hello!")
word = input("Enter something: ")

if word == "hi":
    print("Hi to you too!")
elif word == "hello":
    print("Hello hello!")
elif word == "howdy":
    print("Howdyyyy!")
elif word == "hey":
    print("Hey hey hey!")
else:
    print("I don't know what", word, "means.")
````

# Strings and Lists: indexing 取出其中一個

Before the homework, let's learn how to take out **one item** from a string or a list.

## len(): how long is it?
`len()` tells you how many characters are in a string, or how many items are in a list.

```python
>>> len('hello')
5
>>> len([3, 7, 9, 15])
4
```

## Indexing: get one item by its position
Use square brackets `[ ]` with a position number (called an **index**).
**Python starts counting from 0, not 1!**

```
 word  =   'h'   'e'   'l'   'l'   'o'
 index:     0     1     2     3     4
           -5    -4    -3    -2    -1
```

```python
>>> word = 'hello'
>>> word[0]      # the 1st character
'h'
>>> word[1]      # the 2nd character
'e'
>>> word[-1]     # the last character
'o'
```

Lists work the same way:
```python
>>> nums = [3, 7, 9, 15]
>>> nums[0]
3
>>> nums[3]
15
>>> nums[len(nums) - 1]   # the last item, same as nums[-1]
15
```

## Reversing: [::-1]
Adding `[::-1]` after a string gives you the string **backwards**.
```python
>>> 'hello'[::-1]
'olleh'
```

> **白話說明：**
> - 把字串或串列想成**一排座位**，每個位置都有編號，但編號**從 0 開始**。所以第 1 個是 `[0]`，第 5 個是 `[4]`。
> - 負數是**從後面數**：`[-1]` 是最後一個，`[-2]` 是倒數第二個。
> - 如果有 5 個東西，最後一個的編號是 `5 - 1 = 4`，所以「最後一個」可以寫成 `[len(x) - 1]`。
> - `[::-1]` 先當成固定口訣記起來：「倒過來」。
> - 要找「正中間」那一個，可以用 `len(x) // 2`（長度除以 2 取整數）。

````python
word = 'hello'
print(len(word))
print(word[0])
print(word[-1])
print(word[::-1])

nums = [3, 7, 9, 15, 20]
print(len(nums))
print(nums[0])
print(nums[len(nums) // 2])   # 5 // 2 = 2，所以拿到 nums[2]，也就是 9
````
